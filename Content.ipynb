# Content\-Based Filtering

```
import pandas 
movies = pandas.read_csv("movies_small.csv", sep=';')
```

```
movies
```

```
movies["overview"]
```

```
from sklearn.feature_extraction.text import TfidfVectorizer # Class used to vectorize the texts
tfidf = TfidfVectorizer(stop_words="english")
movies["overview"] = movies["overview"].fillna("") #NaN values will have empty strings as their value
```

```
tfidf_matrix = tfidf.fit_transform(movies["overview"])
pandas.DataFrame(tfidf_matrix.toarray(), columns=tfidf.get_feature_names_out())
# Row = movies, # columns = words
```

```
tfidf_matrix.shape # 121 = no. of words taken out my the "overview" description
```

### Similarity Matrix

```
from sklearn.metrics.pairwise import linear_kernel
similarity_matrix = linear_kernel(tfidf_matrix, tfidf_matrix) # Performs pair-wise analysis
similarity_matrix
```

```
similarity_matrix.shape # 6 by 6 array
```

```
similarity_matrix[1][0] # This means that the second movie [1] has a similarity coefficient of 0 with the first movie [0].
```

You can also see that all the diagonal elements show a complete correlation of 1, since these values represent correlation of movies with themselves\.

### Finding the most similar movies to a certain movie

```
movie_title = "John Carter"
```

Now, as you can see the "John Carter" movie is the 6th movie in the dataset\. Looking at the similarity matrix above for the 6th row, we can see the similarity coefficient of this movie with other movies\. We'll arrange these values in descending order\.

```
sorted(similarity_matrix[5], reverse=True)
```

In practice, we won't know the exact row number of the movie we want to select\.

```
idx = movies.loc[movies["title"]==movie_title].index[0]
idx
```

Let's use an enumerate function to understand which movie does out movie\_title point to\.

```
scores = list(enumerate(similarity_matrix[idx]))
scores
```

```
sorted(scores, reverse=True) # This is wrong, sorts by the first item of the tuple. We need to select the 2nd item.
```

```
scores = sorted(scores, key=lambda x: x[1], reverse=True)
scores
```

```
movies_indices = [i[0] for i in scores[1: 4]] # Includes the first 3 elements, excluding the first one.
movies_indices
```

```
list(movies["title"].iloc[movies_indices]) # Getting the titles of the selected movies
```

Now let's implement this same code into a manually\-defined function for reusability\.

```
def similar_movies(movie_title, nr_movies):
    idx = movies.loc[movies["title"]==movie_title].index[0]
    scores = list(enumerate(similarity_matrix[idx]))
    scores = sorted(scores, key=lambda x: x[1], reverse=True)
    movies_indices = [i[0] for i in scores[1: nr_movies+1]]
    titles = list(movies["title"].iloc[movies_indices])
    return titles
```

```
similar_movies("Kung Fu Panda 3", 3)
```

This is limited in its functioning accuracy\. If you want more better results, you many have to take other relevant info from the dataset \(for example, genre, keywords, etc\.\)
